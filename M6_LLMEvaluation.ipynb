## LLM Evaluation and Metrics

## Overview

Evaluating LLM outputs is difficult because there is often no single correct answer. Some tasks can be evaluated with standard metrics such as accuracy, BLEU, ROUGE, perplexity, or exact match, while others require human judgment about usefulness, clarity, safety, correctness, or potential harm. Increasingly, LLMs are also used as evaluators, but LLM-as-a-judge approaches introduce their own limitations and biases.

* an LLM-as-a-judge evaluation.
